````python
# Check for noisy data: missing values, duplicates, outliers based on z-score and IQR
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('temiz_veri_final.csv', encoding='utf-8-sig')
print('# Shape:', df.shape)

# Missing values
missing = df.isna().sum().sort_values(ascending=False)
print('\n# Top missing values:')
print(missing[missing>0].head(20))

# Duplicates
dups = df.duplicated().sum()
print('\n# Duplicate rows:', dups)

# Numeric columns
num = df.select_dtypes(include=['int64','float64'])

# Outliers via IQR for numeric cols
iqr_outliers = {}
for c in num.columns:
    col = df[c].dropna()
    q1 = col.quantile(0.25)
    q3 = col.quantile(0.75)
    iqr = q3 - q1
    if iqr==0:
        continue
    lower = q1 - 1.5*iqr
    upper = q3 + 1.5*iqr
    out_pct = ((col < lower) | (col > upper)).mean()
    if out_pct>0:
        iqr_outliers[c] = out_pct

print('\n# Columns with IQR outlier proportion >0:')
for k,v in sorted(iqr_outliers.items(), key=lambda x:-x[1])[:20]:
    print(k, round(v,4))

# Extreme z-score outliers (>4)
z_outliers = {}
for c in num.columns:
    col = df[c].dropna()
    if col.std()==0:
        continue
    z = np.abs(stats.zscore(col))
    z_outliers[c] = (z>4).mean()

print('\n# Proportion of extreme z>4 outliers (top 20):')
for k,v in sorted(z_outliers.items(), key=lambda x:-x[1])[:20]:
    if v>0:
        print(k, round(v,6))

# Check categorical anomalies: departments, statuses
print('\n# Unique Delivery Status values:')
print(df['Delivery Status'].value_counts(dropna=False).to_dict() if 'Delivery Status' in df.columns else 'no column')

# Check improbable values: negative prices, negative quantities
if 'Order Item Product Price' in df.columns:
    neg_prices = (df['Order Item Product Price']<0).sum()
    print('\nNegative product prices:', neg_prices)
if 'Order Item Quantity' in df.columns:
    neg_qty = (df['Order Item Quantity']<0).sum()
    print('Negative quantities:', neg_qty)

# Show rows with missing customer id or order id
for c in ['Order Id','Customer Id']:
    if c in df.columns:
        print('\nMissing', c, (df[c].isna()).sum())

# Print a small sample of likely noisy rows: where Sales extremely high or negative
if 'Sales' in df.columns:
    s = df['Sales']
    print('\nSales min, max, mean:', s.min(), s.max(), round(s.mean(),2))
    print('\nSample extreme negative sales:')
    print(df[df['Sales']<s.quantile(0.01)].head(5))
    print('\nSample extreme high sales:')
    print(df[df['Sales']>s.quantile(0.99)].head(5))

````
